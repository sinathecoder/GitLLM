```python
import chromadb
c = chromadb.PersistentClient(path="./git_history_db")   # or "./git_history_api_db"


```

```python
col = c.get_collection("git_history")
print("docs:", col.count())
```

```python
# first N docs + metadata (no embeddings)
print(col.peek(limit=5).get("documents"))

# all documents + metadata
res = col.get(include=["documents", "metadatas"])
for i, doc in enumerate(res["documents"]):
    print(i, res["metadatas"][i], doc[:120])
```
